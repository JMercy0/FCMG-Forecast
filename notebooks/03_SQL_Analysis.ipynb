```python
# ==========================================
# Import Libraries
# ==========================================

import os
import pandas as pd

from dotenv import load_dotenv
from sqlalchemy import create_engine

# ==========================================
# Load Environment Variables
# ==========================================

load_dotenv()

# ==========================================
# PostgreSQL Connection Details
# ==========================================

USERNAME = os.getenv("DB_USER")
PASSWORD = os.getenv("DB_PASSWORD")
HOST = os.getenv("DB_HOST")
PORT = os.getenv("DB_PORT")
DATABASE = os.getenv("DB_NAME")

DATABASE_URL = (
    f"postgresql://{USERNAME}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
)

engine = create_engine(DATABASE_URL)

print("✅ Database Connected Successfully!")
```

```python
query = """
SELECT
    is_weekend,
    ROUND(SUM(net_sales)::numeric,2) AS revenue
FROM fmcg_sales
GROUP BY is_weekend
ORDER BY is_weekend;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT

sku_name,

SUM(net_sales) AS revenue,

RANK() OVER(

ORDER BY SUM(net_sales) DESC

) AS revenue_rank

FROM fmcg_sales

GROUP BY sku_name;
"""

ranking = pd.read_sql(query, engine)
ranking.head(20)
```

```python
query = """
SELECT

sku_name,

SUM(net_sales) revenue,

DENSE_RANK() OVER(

ORDER BY SUM(net_sales) DESC

) rank

FROM fmcg_sales

GROUP BY sku_name;
"""

dense_rank = pd.read_sql(query, engine)
dense_rank.head()
```

```python
query = """
SELECT

date,

SUM(net_sales) AS revenue,

SUM(SUM(net_sales)) OVER(

ORDER BY date

) AS cumulative_revenue

FROM fmcg_sales

GROUP BY date

ORDER BY date;
"""

running_sales = pd.read_sql(query, engine)
running_sales
```

```python
query = """
SELECT
    country,
    SUM(net_sales) AS revenue,
    ROUND(
        (
            100 * SUM(net_sales)
            / SUM(SUM(net_sales)) OVER ()
        )::numeric,
        2
    ) AS revenue_percent
FROM fmcg_sales
GROUP BY country
ORDER BY revenue_percent DESC;
"""

country_share = pd.read_sql(query, engine)
country_share
```
